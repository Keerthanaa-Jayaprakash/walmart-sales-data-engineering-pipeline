```python
# Inspect Excel workbook sheets

import pandas as pd

file_path = "/Volumes/walmart_retail/sales/raw_data/walmart_Retail_Data.xlsx"

excel_file = pd.ExcelFile(file_path)

excel_file.sheet_names
```

```python
# Read the Walmart retail data

df_pandas = pd.read_excel(
    file_path,
    sheet_name="walmart Retail Data"
)

print("Rows:", len(df_pandas))
print("Columns:", len(df_pandas.columns))

display(df_pandas.head(10))
```

```python
# Inspect source data types and missing values

print("=== DATA TYPES ===")
print(df_pandas.dtypes)

print("\n=== MISSING VALUES ===")
print(df_pandas.isnull().sum())

print("\n=== DUPLICATE ROWS ===")
print(df_pandas.duplicated().sum())
```

```python
# Basic dataset statistics

print("Rows:", len(df_pandas))
print("Columns:", len(df_pandas.columns))

print("\nColumn names:")
print(df_pandas.columns.tolist())

print("\nMissing values:")
print(df_pandas.isnull().sum())
```

```python
# Prepare the raw data for Spark ingestion

from pyspark.sql import functions as F

# Convert Pandas object columns to strings while preserving missing values
df_for_spark = df_pandas.copy()

object_columns = df_for_spark.select_dtypes(include=["object"]).columns

for col in object_columns:
    df_for_spark[col] = df_for_spark[col].where(
        df_for_spark[col].notna(),
        None
    )
    df_for_spark[col] = df_for_spark[col].astype("string")

print("Text columns normalized:", list(object_columns))
```

```python
# Convert the prepared Pandas DataFrame to a Spark DataFrame

df_bronze = spark.createDataFrame(df_for_spark)

print("Bronze rows:", df_bronze.count())
print("Bronze columns:", len(df_bronze.columns))

display(df_bronze.limit(10))
```

```python
# Standardize column names for Delta Lake

import re

def clean_column_name(column_name):
    column_name = column_name.strip().lower()
    column_name = re.sub(r"[^a-zA-Z0-9]+", "_", column_name)
    column_name = column_name.strip("_")
    return column_name

clean_columns = [
    clean_column_name(col)
    for col in df_bronze.columns
]

df_bronze_clean = df_bronze.toDF(*clean_columns)

print("Bronze columns:")
print(df_bronze_clean.columns)
```

```python
#Verify the data - to see if any rows/columns has been deleted

print("Rows:", df_bronze_clean.count())
print("Columns:", len(df_bronze_clean.columns))

display(df_bronze_clean.limit(10))
```

```python
# Write Bronze data to Delta Lake

bronze_table = "walmart_retail.sales.bronze_sales"

(
    df_bronze_clean.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable(bronze_table)
)

print(f"Bronze table created: {bronze_table}")
```

```python
bronze_df = spark.table("walmart_retail.sales.bronze_sales")

print("Rows:", bronze_df.count())
print("Columns:", len(bronze_df.columns))

display(bronze_df.limit(10))
```
